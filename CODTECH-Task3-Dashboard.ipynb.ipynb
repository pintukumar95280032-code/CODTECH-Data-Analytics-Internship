```python
import pandas as pd
import plotly.express as px
import dash
from dash import dcc, html

# 1. Loading the housing dataset for analysis
df = pd.read_csv('USA_Housing.csv')

# 2. Creating Interactive Visualizations as per internship requirements

# Generating a Histogram to visualize the distribution of House Prices
fig_hist = px.histogram(df, x="Price", title="Distribution of House Prices",
                        nbins=50, color_discrete_sequence=['#2ecc71'])

# Generating a Scatter Plot to analyze the relationship between Area Income and House Price
fig_scatter = px.scatter(df, x="Avg. Area Income", y="Price",
                         title="Impact of Area Income on House Price",
                         trendline="ols", color_discrete_sequence=['#e74c3c'])

# 3. Initializing the Dash Application
app = dash.Dash(__name__)

# 4. Defining the Dashboard Layout with interactive components
app.layout = html.Div(style={'fontFamily': 'Arial', 'padding': '20px'}, children=[
    html.H1("CODTECH INTERNSHIP: HOUSING ANALYTICS DASHBOARD",
            style={'textAlign': 'center', 'color': '#34495e'}),

    html.Hr(), # Adding a horizontal line for clear sectioning

    # Section for Price Distribution Graph
    html.Div([
        html.H3("1. Price Range Analysis"),
        dcc.Graph(figure=fig_hist)
    ], style={'padding': '10px', 'backgroundColor': '#f9f9f9', 'marginBottom': '20px'}),

    # Section for Income & Price Correlation Graph
    html.Div([
        html.H3("2. Income & Price Correlation"),
        dcc.Graph(figure=fig_scatter)
    ], style={'padding': '10px', 'backgroundColor': '#f9f9f9'})
])

# 5. Running the dashboard directly in the Google Colab environment
if __name__ == '__main__':
    import threading
    from google.colab import output

    # Running the server in a separate thread to prevent blocking the notebook
    thread = threading.Thread(target=app.run, kwargs={'port': 8050})
    thread.start()

    # Rendering the dashboard as an IFrame for live interaction
    output.serve_kernel_port_as_iframe(8050, height=800)
```
